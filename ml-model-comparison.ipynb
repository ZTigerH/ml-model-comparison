```python
import time
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
import numpy as np

soybean = pd.read_csv('soybean.csv')
iris = pd.read_csv('iris.csv', sep='\t')
skin = pd.read_csv('Skin_NonSkin.tsv', sep='\t')
bankruptcy = pd.read_csv('bankruptcy.csv')

datasets = {
    'Soybean': (soybean, 'Label'),
    'Iris': (iris, 'Species'),
    'Skin Segmentation': (skin, 'Label'),
    'Bankruptcy': (bankruptcy, 'Bankrupt?')
}

def evaluate_model(model, X, y, cv=5):
    start_time = time.time()
    scores = cross_val_score(model, X, y, cv=cv, n_jobs=-1)
    end_time = time.time()
    return scores.mean(), end_time - start_time

results = {}

for name, (df, label_col) in datasets.items():
    print(f"Evaluating dataset: {name}")
    
    X = df.drop(columns=[label_col])
    y = df[label_col]
    
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)
    
    results[name] = {}

    # Logistic Regression
    model = LogisticRegression(max_iter=1000)
    accuracy, runtime = evaluate_model(model, X_scaled, y)
    results[name]['Logistic Regression'] = (accuracy, runtime)
    
    # K-Nearest Neighbors
    knn_results = {}
    for metric in ['euclidean', 'manhattan']:
        for k in [5, 10, 15, 20, 25, 30, 35, 40, 45, 50]:
            model = KNeighborsClassifier(n_neighbors=k, metric=metric)
            accuracy, runtime = evaluate_model(model, X_scaled, y)
            knn_results[f'KNN (k={k}, {metric})'] = (accuracy, runtime)
    results[name]['K-Nearest Neighbors'] = knn_results

    # Naive Bayes
    model = GaussianNB()
    accuracy, runtime = evaluate_model(model, X, y)
    results[name]['Naive Bayes'] = (accuracy, runtime)
    
    # Decision Trees
    dt_results = {}
    for max_depth in [5, 10, 15, 20, 25]:
        model = DecisionTreeClassifier(criterion='entropy', max_depth=max_depth)
        accuracy, runtime = evaluate_model(model, X, y)
        dt_results[f'Decision Tree (max_depth={max_depth})'] = (accuracy, runtime)
    results[name]['Decision Trees'] = dt_results

    # Random Forests
    rf_results = {}
    for n_estimators in [10, 50, 100]:  # Adjust the values as discussed in class
        model = RandomForestClassifier(criterion='entropy', n_estimators=n_estimators)
        accuracy, runtime = evaluate_model(model, X, y)
        rf_results[f'Random Forest (n_estimators={n_estimators})'] = (accuracy, runtime)
    results[name]['Random Forests'] = rf_results

    # SVM
    svm_results = {}
    for kernel in ['linear', 'rbf']:
        model = SVC(kernel=kernel)
        accuracy, runtime = evaluate_model(model, X_scaled, y)
        svm_results[f'SVM ({kernel})'] = (accuracy, runtime)
    results[name]['SVMs'] = svm_results

for dataset, models in results.items():
    print(f"\nResults for {dataset} Dataset:")
    for model_name, result in models.items():
        if isinstance(result, dict):
            for sub_model, (accuracy, runtime) in result.items():
                print(f"{sub_model}: Accuracy = {accuracy:.4f}, Runtime = {runtime:.4f} seconds")
        else:
            accuracy, runtime = result
            print(f"{model_name}: Accuracy = {accuracy:.4f}, Runtime = {runtime:.4f} seconds")
```
